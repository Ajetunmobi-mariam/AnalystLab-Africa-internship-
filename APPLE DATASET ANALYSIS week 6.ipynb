# Feature Engineering

### Daily price change

```python
df['Daily_Price_Change'] = df['Close'] - df['Open']    
```

```python
df[['Open', 'Close', 'Daily_Price_Change']].head()
```

### 7 days moving Average

```python
df['MA_7'] = df['Close'].rolling(window=7).mean()             
```

### 30 days moving Average

```python
df['MA_30'] = df['Close'].rolling(window=30).mean()           
```

```python
df[['Close', 'MA_7', 'MA_30']].tail()
```

### Percentage price change

```python
df['Pct_Price_Change'] = ((df['Close'] - df['Open']) / df['Open']) * 100          
```

```python
df[['Pct_Price_Change']].head()
```

### Daily Return

```python
df['Daily_Return'] = df['Close'].pct_change()           
```

### Monthly Return

```python
monthly_returns = df['Close'].resample('ME').last().pct_change()            
```

#### Volatility measures

```python
df['Volatility_30D'] = df['Daily_Return'].rolling(30).std()
```

```python
df[['Daily_Return', 'Volatility_30D']].tail()
```

#### 3.Daily and Monthly performance Analysis

###### BEST AND WORSE MONTH PERFORMANCE

```python
monthly_returns.idxmax()
```

```python
best_month = monthly_returns.idxmax()
best_return = monthly_returns.max()

print("Best Month:", best_month.strftime('%B %Y'))
print("Monthly Return:", round(best_return * 100, 2), "%")

worst_month = monthly_returns.idxmin()
worst_return = monthly_returns.min()

print("Worst Month:", worst_month.strftime('%B %Y'))
print("Monthly Return:", round(worst_return * 100, 2), "%")
```

#### WORST MONTH

###### BEST AND WORSE DAILY RETURN PERFORMANCE

```python
best_day = df['Daily_Return'].idxmax()
best_return = df['Daily_Return'].max()

worst_day = df['Daily_Return'].idxmin()
worst_return = df['Daily_Return'].min()

print("Best Daily Performance")
print("Date:", best_day.strftime('%d %B %Y'))
print("Daily Return:", round(best_return * 100, 2), "%")

print("\nWorst Daily Performance")
print("Date:", worst_day.strftime('%d %B %Y'))
print("Daily Return:", round(worst_return * 100, 2), "%")
```

# Visualization

### 1. Stock Closing price trend

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(df.index, df['Close'], color='blue', linewidth=2)

plt.title('Apple Stock Closing Price Trend')
plt.xlabel('Date')
plt.ylabel('Closing Price ($)')
plt.grid(True)

plt.show()
```

### 2.Trading volume Trend

```python
plt.figure(figsize=(12,6))
plt.plot(df.index, df['Volume'], color='purple')

plt.title('Apple Trading Volume Trend')
plt.xlabel('Date')
plt.ylabel('Trading Volume')
plt.grid(True)

plt.show()
```

### 3.Moving Average Analysis

```python
plt.figure(figsize=(12,6))

plt.plot(df.index, df['Close'],
         label='Close Price',
         color='blue')

plt.plot(df.index, df['MA_7'],
         label='7-Day Moving Average',
         color='orange')

plt.plot(df.index, df['MA_30'],
         label='30-Day Moving Average',
         color='red')

plt.title('Moving Average Analysis')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)

plt.show()
```

### 4.Monthly return

```python
plt.figure(figsize=(12,6))

plt.plot(monthly_returns.index,
         monthly_returns * 100,
         color='brown')

plt.title('Monthly Returns Analysis')
plt.xlabel('Date')
plt.ylabel('Monthly Return (%)')
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

plt.figure(figsize=(12,6))

plt.plot(monthly_returns.index,
         monthly_returns * 100,
         color='brown',
         linewidth=2)

plt.title('Monthly Returns Analysis')
plt.xlabel('Month')
plt.ylabel('Monthly Return (%)')

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.gca().xaxis.set_major_locator(mdates.MonthLocator(interval=3))

plt.xticks(rotation=45)

plt.grid(True)
plt.show()
```

### 5. Volatility Trend

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.plot(df.index,
         df['Volatility_30D'] * 100,
         color='darkred',
         linewidth=2)

plt.title(' Volatility Trend')
plt.xlabel('Date')
plt.ylabel('Volatility (%)')

plt.grid(True)
plt.show()
```

# Key findings
1. Apple's stock price showed an overall upward trend throughout the five-year period, demonstrating strong long-term growth.
The most significant growth occurred between 2023 and 2026, despite periods of market fluctuations.

2. Trading volume varied throughout the analysis period, with the highest trading activity recorded in late 2024, indicating increased market participation.

3. The 7-day and 30-day moving averages confirmed the stock's positive long-term trend while smoothing short-term price fluctuations.

4. Monthly returns fluctuated considerably, with both strong positive and negative performance periods observed.

5. A period of weaker performance was observed between September and December 2022, where monthly returns remained mostly negative.

6. Volatility levels changed over time, with the highest volatility occurring around mid-2025, indicating increased market uncertainty.



# Conclusion

This analysis examined Apple's stock performance over a five-year period using time-series analysis techniques, including trend analysis, return analysis, moving averages, and volatility measurement. The findings revealed that Apple experienced strong long-term growth despite short-term fluctuations and periods of market uncertainty. Overall, the stock demonstrated resilience, sustained growth, and positive performance, making it a valuable case study for understanding stock market behavior.
